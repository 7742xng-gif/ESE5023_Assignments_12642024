```python
import numpy as np 
M1= np.random.randint(0, 51, (5,10))
print(M1)
M2=np.random.randint(0,51,(10,5))
print(M2)


def Matrix_multip():
    
    M3 = np.ones((5, 5), dtype=int)

    for i in range(5):
        for j in range(5):
            
            x = 0
            
            for k in range(10):
                x = x + M1[i][k] * M2[k][j]
            
            M3[i][j] = x

    print(M3)
Matrix_multip()
```

```python
def Pascal_triangle(k):
    row = [1]
    for i in range(k):
        new_row = [1] * (len(row) + 1)
        for j in range(len(row) - 1):
            x = row[j] + row[j + 1]
            new_row[j + 1] = x
        row = new_row
    print(row)

Pascal_triangle(100)
Pascal_triangle(200)
```

```python
def Least_moves(x):
    moves = 0

    for i in range(x):
        if x <= 1:
            break

        if x % 2 == 0:
            x = x // 2
        else:
            x = x - 1

        moves = moves + 1

    print(moves)
Least_moves(2)
Least_moves(5)
Least_moves(19)
```

```python
def Find_expression(x):
    digits = 123456789
    count = 0
    for i in range(6561):

        expression = "1"
        a = i
        total = 0
        num = 1
        signs = 1
        for j in range(8):
            r = a % 3
            digit = digits // (10 ** (7-j)) % 10
            if r == 0:
                sign = ""
                num = num * 10 + digit
            elif r == 1:
                sign = "+"
                total = total + signs * num
                num = digit
                signs = 1
            else:
                sign = "-"
                total = total + signs * num
                num = digit
                signs = -1
            expression = expression + sign + str(digit)
            a = a // 3
        total = total + signs * num
        if total == x:
            count = count + 1
            print(expression, "=", x)
    return count

Find_expression(50)
Find_expression(12)
```

```python
import matplotlib.pyplot as plt
Total_solutions = [0] * 100
for i in range(1, 101):
    count = Find_expression(i)
    Total_solutions[i-1] = count
plt.plot(range(1, 101), Total_solutions)

plt.xlabel("Number")
plt.ylabel("Total solutions")
plt.title("Total Solutions for Numbers 1 to 100")
plt.show()

print("Maximum:", max(Total_solutions))
print("Minimum:", min(Total_solutions))

for i in range(100):
    if Total_solutions[i] == max(Total_solutions):
        print("Maximum number:", i + 1)
    if Total_solutions[i] == min(Total_solutions):
        print("Minimum number:", i + 1)
```
